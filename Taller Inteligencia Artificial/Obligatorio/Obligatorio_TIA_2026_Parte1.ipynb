## Preprocesando highway-env

Replicamos el pipeline de observación del paper original de DQN (Mnih et al., 2013) sobre highway-env. La consigna es deliberada: el entorno por defecto entrega una observación estructurada (`KinematicObservation`: matriz $V \times F$ con posiciones, velocidades y atributos de los $V$ vehículos cercanos), que es esencialmente un *oráculo simbólico*. Si entrenáramos sobre eso, el problema se reduciría a un MLP pequeño y convergería en pocos miles de pasos — pero el agente nunca aprendería **percepción**. Forzar la entrada visual nos pone en el mismo régimen que Atari: la red tiene que aprender features visuales (carriles, vehículos, geometría) **además de** la política.

El pipeline está armado con wrappers de gymnasium estándar. La función $\phi: \mathcal{O} \to \mathbb{R}^{4 \times H \times W}$ aplica:

#### 1. Render como observación

Reemplazamos la observación nativa con el frame RGB devuelto por `env.render(mode="rgb_array")`. El wrapper `_RenderAsObservation` (en `utils.py`) intercepta cada `step()` y descarta la `KinematicObservation`. Costo marginal: una llamada extra a `render()` por step; highway ya rendea internamente para el video, así que reusa el mismo path de pygame.

#### 2. Resize $600 \times 150 \to 336 \times 84$ (W × H)

El render nativo es $600 \times 150$ (aspect 4:1, autopista landscape). Mnih et al. usaron $84 \times 84$ para Atari, donde la pantalla nativa es $210 \times 160$. 
Conservar aspect en $336 \times 84$ (W × H, sigue siendo 4:1) es un compromiso: preserva la geometría del scene a costa de inflar la primera capa fully-connected.

> Se deja a decisión de cada equipo si usar crops y/o resize siempre y cuando este debidamente justificado. 

#### 3. Conversión a escala de grises

Aplicamos la fórmula de luma estándar **ITU-R BT.601** (Recomendación de la International Telecommunication Union, 1982, para codificación de TV digital):

$$Y' = 0.299\,R + 0.587\,G + 0.114\,B$$

Los pesos derivan de la sensibilidad fotópica del ojo humano: más sensible al verde, después al rojo, y mucho menos al azul. Es el default de `gymnasium.wrappers.GrayscaleObservation` (que internamente usa estos coeficientes; ver también `cv2.cvtColor(..., COLOR_RGB2GRAY)` y `skimage.color.rgb2gray`, que implementan la misma fórmula).

En highway, el color carga poca información útil para conducción: carriles blancos sobre gris, vehículos de colores arbitrarios que no codifican intención. Reduce el input 3× sin pérdida de señal.

#### 4. Frame stacking: $1 \to 4$ canales temporales

Ver solo un frame no contiene velocidad ni aceleración. La stack $\phi_t = (s_{t-3}, s_{t-2}, s_{t-1}, s_t)$ permite a la red inferir derivadas temporales por diferencia entre frames consecutivos. Es el truco estándar de Mnih.

`FrameStackObservation` mantiene una ventana deslizante de tamaño 4: dos observaciones consecutivas $\phi_t$ y $\phi_{t+1}$ comparten 3 frames.


```python
OBS_HEIGHT = 84  # alto de la imagen de entrada a la red
OBS_WIDTH = 336  # ancho — preserva aspect 4:1 del render original
NUM_STACKED_FRAMES = 4  # apilamos los últimos 4 frames
```

### `simulation_frequency` vs `policy_frequency`: el frame-skip de highway

A diferencia de Atari, highway-env no usa un wrapper `AtariPreprocessing(frame_skip=k)`. La dinámica equivalente vive en el config del env:

* $f_\pi$ = `policy_frequency`: tasa de decisión del agente (Hz, **en tiempo simulado**).
* $f_s$ = `simulation_frequency`: tasa de integración del simulador físico (IDM + cinemática bicycle) (Hz, **en tiempo simulado**).

Con $f_\pi = 1$ y $f_s = 5$, cada `env.step()` ejecuta $k = f_s / f_\pi = 5$ sub-pasos de simulación manteniendo constante la última acción, y devuelve el frame final. **Es algorítmicamente idéntico** al `frame_skip = k` de `AtariPreprocessing` (action repeat $k$ + observación al $k$-ésimo frame). La diferencia es de implementación: highway lo hace dentro de su `_simulate()` en vez de un wrapper externo.

#### Tiempo simulado vs tiempo real

Dos relojes desacoplados conviven en cualquier simulador:

| Reloj | Definición | Ejemplo |
|---|---|---|
| Simulado | Tiempo *dentro* del mundo del juego | 30 s de tráfico en la autopista |
| Real (wall-clock) | Tiempo de ejecución del proceso | Esos 30 s simulados pueden tardar 0.5 s reales en GPU |

`simulation_frequency` mide pasos de física **por segundo simulado**, no por segundo real. Subirlo de 5 → 30 multiplica por 6 los sub-pasos por `env.step()`: física más precisa + 6× más frames capturados para el video (vía el hook `set_record_video_wrapper`), a costa de 6× más wall-time por step.

#### Por qué dos valores: `SIM_FREQUENCY_TRAIN` ≠ `SIM_FREQUENCY_EVAL`

| Constante | Valor | Justificación |
|---|---|---|
| `SIM_FREQUENCY_TRAIN` | 5 | Wall-time es el cuello de botella |
| `SIM_FREQUENCY_EVAL` | 30 | Pocos episodios, wall-time despreciable. Necesitamos videos fluidos para entregar. |

Los videos que `RecordVideo` graba **durante** el training (cada 50 episodios) van a quedar a 5 fps (perceptiblemente choppy). Sirven para monitoreo de progreso, no como demos finales.

> Implementación completa en `make_env` (`utils.py`).


```python
# Hz de física integrados por segundo simulado. Es independiente de
# policy_frequency (= 1 Hz por default = el agente decide 1 vez por segundo
# simulado). Subir SIM_FREQUENCY hace que cada env.step() simule mas
# sub-pasos de fisica → mas frames capturados para el video → mas suave,
# pero entrenamiento mas lento (proporcionalmente).
SIM_FREQUENCY_TRAIN = 5  # rapido para entrenar muchos steps
SIM_FREQUENCY_EVAL = 30  # smooth para los videos finales
```

```python
env = make_env(
    ENV_NAME,
    video_folder="videos/random_agent",
    record_every=1,  # grabamos video de cada episodio
    obs_height=OBS_HEIGHT,
    obs_width=OBS_WIDTH,
    stack_frames=NUM_STACKED_FRAMES,
    simulation_frequency=SIM_FREQUENCY_EVAL,
    seed=SEED,
    duration=EPISODE_DURATION,
    lanes_count=LANES_COUNT,
    vehicles_count=VEHICLES_COUNT,
    vehicles_density=VEHICLES_DENSITY,
    high_speed_reward=HIGH_SPEED_REWARD,
)
```

```python
# generamos un episodio de ejemplo para mostrar el entorno
obs, info = env.reset(seed=SEED)
first_time = True
done = False
while not done:
    action = env.action_space.sample()  # accion aleatoria
    obs, reward, terminated, truncated, info = env.step(action)
    if first_time:
        show_observation_stack(obs)
        first_time = False
    done = terminated or truncated
```

# Deep Q Learning

Deep Q Learning extiende el algoritmo clásico de Q-learning al emplear una **red neuronal profunda** como aproximador de la función de valor $Q(s,a)$. Inspirado en Mnih et al. (2013), este método utiliza una **red convolucional** para procesar directamente las imágenes del entorno Atari, un **replay buffer** para romper la correlación temporal de las muestras. La política sigue un esquema **ε-greedy**, balanceando exploración y explotación, y se entrena minimizando el error de la ecuación de Bellman sobre lotes de transiciones muestreadas de manera aleatoria.

## Memoria

El algoritmo de Deep Q Learning presentado en el paper utiliza una memoria (llamada Replay Memory) para almacenar transiciones pasadas. Tuplas que contienen un estado base, la accion tomada, la recompensa obtenida, una bandera que indica si el siguiente estado es final o no; y el estado siguiente.

Esta memoria es circular, es decir, tiene un límite maximo de elementos y una vez esté llena comienza a reemplazar los elementos más viejos.

Vamos a necesitar crear una función **sample** que obtiene una mustra aleatoria de elementos de la memoria.  Esto puede ser una lista de Transiciones o listas separadas (pero alineadas) de los elementos que las componen.


> Para implementar esta funcionalidad se debe modificar el archivo **replay_memory.py**

```python
from replay_memory import ReplayMemory, Transition

# Creamos la memoria de repetición
replay_memory = ReplayMemory(3)
tran1 = Transition("state1", "action1", "reward1", "term1", "next_state1")
tran2 = Transition("state2", "action2", "reward2", "term2", "next_state2")
tran3 = Transition("state3", "action3", "reward3", "term3", "next_state3")
tran4 = Transition("state4", "action4", "reward4", "term4", "next_state4")

# Añadimos transiciones a la memoria (solo las 3 últimas se guardan)
replay_memory.add(tran1.state, tran1.action, tran1.reward, tran1.terminated, tran1.next_state)
replay_memory.add(tran2.state, tran2.action, tran2.reward, tran2.terminated, tran2.next_state)
replay_memory.add(tran3.state, tran3.action, tran3.reward, tran3.terminated, tran3.next_state)
replay_memory.add(tran4.state, tran4.action, tran4.reward, tran4.terminated, tran4.next_state)

# Mostramos el tamaño de la memoria
print(f"Memory size: {len(replay_memory)}\n")

# Mostramos un sample de la memoria
sampled = replay_memory.sample(2)
print(f"Memory sample:")
for i, sample in enumerate(sampled):
    print(f"Sample {i}: {sample}")
```

## Phi 

La función para procesar los estados (phi en el paper) que es necesaria para poder usar el modelo de Pytorch con las representaciones de gym. Esta función pasa una observación de gym a un tensor de Pytorch y la normaliza.

> Técnicamente la función phi tiene más responsabilidades, como la de apilar los frames y el downsampling. En nuestro caso se lo delegamos a los wrappers de gymnasium.

```python
def process_state(obs):
    """Convierte obs(s) uint8 numpy -> float32 tensor [0,1] (no cambia de device).

    Acepta dos formas, asi greedy_action (1 obs) y _sample_batch (batch)
    pueden compartir la misma funcion sin duplicar la normalizacion:
      - obs.shape == (C, H, W)        -> retorna (1, C, H, W)
      - obs.shape == (B, C, H, W)     -> retorna (B, C, H, W)
    """
    arr = np.asarray(obs, dtype=np.uint8)
    t = torch.from_numpy(arr).float().div_(255.0)
    # La obs (post wrappers) es 3D (stack, H, W). Si nos llega 3D es una
    # sola obs y le agregamos el batch dim; si es 4D ya viene batcheada
    # desde _sample_batch y la dejamos como esta.
    if t.ndim == 3:
        t = t.unsqueeze(0)
    return t


obs, _ = env.reset()
obs_tensor = process_state(obs)
print(f"Observation tensor shape: {obs_tensor.shape}, dtype: {obs_tensor.dtype}")
```

## Entrenamiento

```python
# Hiperparámetros de entrenamiento del agente DQN
#
# Nota: estos valores son un piso para tener una corrida razonable en tiempo
# acotado (notebook ORT / Colab T4, ~1-2 h por agente).
TOTAL_STEPS = 100_000 # tope total de pasos de entrenamiento (subir si el reward se estanca)
EPISODES = 2_000 # tope de episodios (si se alcanza antes de TOTAL_STEPS, termina igual)
STEPS_PER_EPISODE = 1_000 # maximo de pasos por episodio (si se alcanza antes de done, termina igual)

EPSILON_INI = 1.0 # valor inicial de epsilon
EPSILON_MIN = 0.05 # valor minimo de epsilon
EPSILON_ANNEAL_STEPS = 20_000 # pasos para anneal epsilon

EPISODE_BLOCK = 50  # ventana para el reward promedio en tqdm

BATCH_SIZE = 32 # cantidad de transiciones a samplear del replay memory para cada update
BUFFER_SIZE = 10_000 # capacidad del replay memory (cantidad maxima de transiciones que puede almacenar)

GAMMA = 0.99
LEARNING_RATE = 5e-4 # tasa de aprendizaje para el optimizador del modelo DQN

LEARNING_STARTS = 1_000  # warmup: no entrenamos hasta tener 1k transiciones

TRAINING_RECORD_EVERY = 200  # grabamos video de cada N episodios durante el entrenamiento
```

```python
from dqn_agent import DQNAgent

env = make_env(
    ENV_NAME,
    video_folder="videos/dqn_training",
    record_every=TRAINING_RECORD_EVERY,  
    obs_height=OBS_HEIGHT,
    obs_width=OBS_WIDTH,
    stack_frames=NUM_STACKED_FRAMES,
    simulation_frequency=SIM_FREQUENCY_TRAIN,
    seed=SEED,
    duration=EPISODE_DURATION,
    lanes_count=LANES_COUNT,
    vehicles_count=VEHICLES_COUNT,
    vehicles_density=VEHICLES_DENSITY,
    high_speed_reward=HIGH_SPEED_REWARD,
)

net = DQN_CNN_Model(env.observation_space.shape, env.action_space.n).to(DEVICE)

dqn_agent = DQNAgent(
    env,
    net,
    process_state,
    memory_buffer_size=BUFFER_SIZE,
    batch_size=BATCH_SIZE,
    learning_rate=LEARNING_RATE,
    gamma=GAMMA,
    epsilon_i=EPSILON_INI,
    epsilon_f=EPSILON_MIN,
    epsilon_anneal_steps=EPSILON_ANNEAL_STEPS,
    episode_block=EPISODE_BLOCK,
    device=DEVICE,
    learning_starts=LEARNING_STARTS,
    checkpoint_path="dqn_highway.dat",
)

dqn_rewards = dqn_agent.train(EPISODES, STEPS_PER_EPISODE, TOTAL_STEPS)

env.close()
```

```python
env = make_env(
    ENV_NAME,
    video_folder="./videos/dqn_validation",
    record_every=1,  # graba cada episodio
    obs_height=OBS_HEIGHT,
    obs_width=OBS_WIDTH,
    stack_frames=NUM_STACKED_FRAMES,
    simulation_frequency=SIM_FREQUENCY_EVAL,
    seed=SEED,
    duration=EPISODE_DURATION,
    lanes_count=LANES_COUNT,
    vehicles_count=VEHICLES_COUNT,
    vehicles_density=VEHICLES_DENSITY,
    high_speed_reward=HIGH_SPEED_REWARD,
)

dqn_agent.play(env, episodes=3)

env.close()
```

# Double Deep Q Learning

Double DQN mejora la versión clásica de DQN corrigiendo el sesgo de sobreestimación de los valores 
𝑄
Q. Para ello, desacopla la selección de la acción de su evaluación: en cada paso, la red online elige la acción que maximiza 
𝑄
Q, pero la red objetivo distinta estima el valor de esa acción. Esta separación reduce el sesgo hacia valores demasiado optimistas y aporta mayor estabilidad al entrenamiento. El resto de la estructura —replay buffer, política ε-greedy, etc— se mantiene igual que en DQN, aprovechando así un diseño casi idéntico al original pero con resultados más fiables .

Fuente: [arXiv:1509.06461](https://arxiv.org/abs/1509.06461)

> Vamos a utilizar el mismo modelo de red neuronal creado para el problema anterior y la misma implementación de memoria, dejamos un esqueleto de un agente de Double Deep Q learning para completar en el archivo **double_dqn_agent.py**

```python
SYNC_TARGET = 1_000  # cada N updates sincronizamos target_net (DDQN)
```

```python
from double_dqn_agent import DoubleDQNAgent

env = make_env(
    ENV_NAME,
    video_folder="videos/ddqn_training",
    record_every=TRAINING_RECORD_EVERY,
    obs_height=OBS_HEIGHT,
    obs_width=OBS_WIDTH,
    stack_frames=NUM_STACKED_FRAMES,
    simulation_frequency=SIM_FREQUENCY_TRAIN,
    seed=SEED,
    duration=EPISODE_DURATION,
    lanes_count=LANES_COUNT,
    vehicles_count=VEHICLES_COUNT,
    vehicles_density=VEHICLES_DENSITY,
    high_speed_reward=HIGH_SPEED_REWARD,
)

modelo_a = DQN_CNN_Model(env.observation_space.shape, env.action_space.n).to(DEVICE)
modelo_b = DQN_CNN_Model(env.observation_space.shape, env.action_space.n).to(DEVICE)

ddqn_agent = DoubleDQNAgent(
    env,
    modelo_a,
    modelo_b,
    process_state,
    memory_buffer_size=BUFFER_SIZE,
    batch_size=BATCH_SIZE,
    learning_rate=LEARNING_RATE,
    gamma=GAMMA,
    epsilon_i=EPSILON_INI,
    epsilon_f=EPSILON_MIN,
    epsilon_anneal_steps=EPSILON_ANNEAL_STEPS,
    episode_block=EPISODE_BLOCK,
    device=DEVICE,
    sync_target=SYNC_TARGET,
    learning_starts=LEARNING_STARTS,
    checkpoint_path="ddqn_highway.dat",
)

ddqn_rewards = ddqn_agent.train(EPISODES, STEPS_PER_EPISODE, TOTAL_STEPS)

env.close()
```
