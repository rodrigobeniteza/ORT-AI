```python
# Análisis de distribución de clases

# Función para contar muestras por clase en un dataset
def count_samples_per_class(dataset):
    class_counts = {}
    for _, label in dataset:
        if label in class_counts:
            class_counts[label] += 1
        else:
            class_counts[label] = 1
    return class_counts
```

```python
# Análisis Detallado del Desbalance de Clases

import numpy as np
import pandas as pd
from scipy import stats
import seaborn as sns

def calculate_class_imbalance_metrics(train_counts, val_counts, class_names):
    """
    Calcula múltiples métricas de desbalance de clases
    """
    total_counts = [train_counts[i] + val_counts[i] for i in range(len(class_names))]
    total_samples = sum(total_counts)
    
    # Métricas básicas
    max_count = max(total_counts)
    min_count = min(total_counts)
    mean_count = np.mean(total_counts)
    std_count = np.std(total_counts)
    
    # 1. Imbalance Ratio (IR) - Ratio de la clase mayoritaria a la minoritaria
    imbalance_ratio = max_count / min_count
    
    # 2. Coefficient of Variation (CV)
    cv = std_count / mean_count
    
    # 3. Gini Coefficient - Mide desigualdad (0 = perfectamente balanceado, 1 = máximo desbalance)
    def gini_coefficient(counts):
        n = len(counts)
        mean_count = np.mean(counts)
        sum_diff = sum(abs(counts[i] - counts[j]) for i in range(n) for j in range(n))
        return sum_diff / (2 * n * n * mean_count)
    
    gini = gini_coefficient(total_counts)
    
    # 4. Shannon Entropy - Mide diversidad (mayor valor = más balanceado)
    proportions = np.array(total_counts) / total_samples
    shannon_entropy = -sum(p * np.log2(p) for p in proportions if p > 0)
    max_entropy = np.log2(len(class_names))  # Máxima entropía posible
    normalized_entropy = shannon_entropy / max_entropy
    
    # 5. Chi-square test - Test estadístico para uniformidad
    expected_count = total_samples / len(class_names)
    chi2_stat, chi2_p_value = stats.chisquare(total_counts)
    
    # 6. Minority/Majority ratios por clase
    minority_ratios = [count / max_count for count in total_counts]
    majority_ratios = [count / min_count for count in total_counts]
    
    # 7. Percentage distribution
    percentages = [count / total_samples * 100 for count in total_counts]
    
    # 8. Imbalance severity classification
    def classify_imbalance(ir):
        if ir <= 1.5:
            return "BALANCED", "🟢"
        elif ir <= 3:
            return "MILDLY IMBALANCED", "🟡"
        elif ir <= 10:
            return "MODERATELY IMBALANCED", "🟠"
        else:
            return "SEVERELY IMBALANCED", "🔴"
    
    severity, severity_emoji = classify_imbalance(imbalance_ratio)
    
    return {
        'total_counts': total_counts,
        'percentages': percentages,
        'imbalance_ratio': imbalance_ratio,
        'cv': cv,
        'gini': gini,
        'shannon_entropy': shannon_entropy,
        'normalized_entropy': normalized_entropy,
        'chi2_stat': chi2_stat,
        'chi2_p_value': chi2_p_value,
        'minority_ratios': minority_ratios,
        'majority_ratios': majority_ratios,
        'severity': severity,
        'severity_emoji': severity_emoji,
        'max_count': max_count,
        'min_count': min_count,
        'mean_count': mean_count,
        'std_count': std_count
    }

# Calculamos las métricas de desbalance
imbalance_metrics = calculate_class_imbalance_metrics(train_counts, val_counts, class_names)

# Creamos un DataFrame para mejor visualización
df_balance = pd.DataFrame({
    'Clase': class_names,
    'Train': train_counts,
    'Val': val_counts,
    'Total': imbalance_metrics['total_counts'],
    'Porcentaje': imbalance_metrics['percentages'],
    'Ratio_Minoritaria': imbalance_metrics['minority_ratios'],
    'Ratio_Mayoritaria': imbalance_metrics['majority_ratios']
})

# Ordenamos por total descendente
df_balance = df_balance.sort_values('Total', ascending=False).reset_index(drop=True)

# Visualizaciones
fig, axes = plt.subplots(2, 3, figsize=(20, 12))
fig.suptitle('Análisis Detallado del Desbalance de Clases', fontsize=16, fontweight='bold')

# 1. Gráfico de barras con porcentajes
bars = axes[0, 0].bar(range(len(class_names)), imbalance_metrics['percentages'], 
                      color=['red' if p < 8 else 'orange' if p < 9 else 'lightgreen' if p < 11 else 'green' 
                             for p in imbalance_metrics['percentages']], alpha=0.8)
axes[0, 0].set_title('Distribución Porcentual por Clase')
axes[0, 0].set_xlabel('Clases')
axes[0, 0].set_ylabel('Porcentaje (%)')
axes[0, 0].set_xticks(range(len(class_names)))
axes[0, 0].set_xticklabels(class_names, rotation=45)
axes[0, 0].axhline(y=10, color='blue', linestyle='--', alpha=0.7, label='Balance Ideal (10%)')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Añadir valores en las barras
for i, (bar, pct) in enumerate(zip(bars, imbalance_metrics['percentages'])):
    height = bar.get_height()
    axes[0, 0].text(bar.get_x() + bar.get_width()/2., height + 0.1,
                    f'{pct:.1f}%', ha='center', va='bottom', fontweight='bold')

# 2. Ratios respecto a clase minoritaria
axes[0, 1].bar(range(len(class_names)), imbalance_metrics['majority_ratios'], 
               color='skyblue', alpha=0.8)
axes[0, 1].set_title('Ratio respecto a Clase Minoritaria')
axes[0, 1].set_xlabel('Clases')
axes[0, 1].set_ylabel('Ratio (veces más grande)')
axes[0, 1].set_xticks(range(len(class_names)))
axes[0, 1].set_xticklabels(class_names, rotation=45)
axes[0, 1].axhline(y=1.5, color='orange', linestyle='--', alpha=0.7, label='Umbral Balance Suave')
axes[0, 1].axhline(y=3, color='red', linestyle='--', alpha=0.7, label='Umbral Desbalance Moderado')
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

# 3. Heatmap de comparación Train vs Val
train_val_matrix = np.array([train_counts, val_counts])
im = axes[0, 2].imshow(train_val_matrix, cmap='YlOrRd', aspect='auto')
axes[0, 2].set_title('Heatmap: Train vs Validation')
axes[0, 2].set_xlabel('Clases')
axes[0, 2].set_ylabel('Split')
axes[0, 2].set_xticks(range(len(class_names)))
axes[0, 2].set_xticklabels(class_names, rotation=45)
axes[0, 2].set_yticks([0, 1])
axes[0, 2].set_yticklabels(['Train', 'Val'])

# Añadir valores en el heatmap
for i in range(2):
    for j in range(len(class_names)):
        text = axes[0, 2].text(j, i, int(train_val_matrix[i, j]),
                               ha="center", va="center", color="black", fontweight='bold')

plt.colorbar(im, ax=axes[0, 2])

# 4. Distribución acumulativa
sorted_counts = sorted(imbalance_metrics['total_counts'])
cumulative_pct = np.cumsum(sorted_counts) / sum(sorted_counts) * 100
axes[1, 0].plot(range(1, len(sorted_counts) + 1), cumulative_pct, 'bo-', linewidth=2, markersize=8)
axes[1, 0].set_title('Distribución Acumulativa de Muestras')
axes[1, 0].set_xlabel('Clases (ordenadas por cantidad)')
axes[1, 0].set_ylabel('Porcentaje Acumulativo (%)')
axes[1, 0].grid(alpha=0.3)
axes[1, 0].axhline(y=50, color='red', linestyle='--', alpha=0.7, label='50% de datos')
axes[1, 0].axhline(y=80, color='orange', linestyle='--', alpha=0.7, label='80% de datos')
axes[1, 0].legend()

# 5. Box plot de splits Train/Val por clase
split_data = []
split_labels = []
for i, class_name in enumerate(class_names):
    split_data.extend([train_counts[i], val_counts[i]])
    split_labels.extend([f'{class_name}\nTrain', f'{class_name}\nVal'])

# Reorganizar para box plot
train_data = train_counts
val_data = val_counts
axes[1, 1].boxplot([train_data, val_data], labels=['Train', 'Validation'])
axes[1, 1].set_title('Distribución de Muestras por Split')
axes[1, 1].set_ylabel('Número de Muestras')
axes[1, 1].grid(alpha=0.3)

# 6. Pie chart con top/bottom clases
top_3_indices = np.argsort(imbalance_metrics['total_counts'])[-3:]
bottom_3_indices = np.argsort(imbalance_metrics['total_counts'])[:3]
other_sum = sum(imbalance_metrics['total_counts']) - sum(imbalance_metrics['total_counts'][i] for i in np.concatenate([top_3_indices, bottom_3_indices]))

pie_labels = []
pie_values = []
pie_colors = []

for i, idx in enumerate(top_3_indices):
    pie_labels.append(f"Top {i+1}: {class_names[idx]}")
    pie_values.append(imbalance_metrics['total_counts'][idx])
    pie_colors.append(f"C{i}")

for i, idx in enumerate(bottom_3_indices):
    pie_labels.append(f"Bot {i+1}: {class_names[idx]}")
    pie_values.append(imbalance_metrics['total_counts'][idx])
    pie_colors.append(f"C{i+3}")

if other_sum > 0:
    pie_labels.append("Otras")
    pie_values.append(other_sum)
    pie_colors.append("gray")

axes[1, 2].pie(pie_values, labels=pie_labels, autopct='%1.1f%%', colors=pie_colors)
axes[1, 2].set_title('Top 3 vs Bottom 3 Clases')

plt.tight_layout()
plt.show()

# Reporte detallado
print("=" * 90)
print("🔍 ANÁLISIS DETALLADO DEL DESBALANCE DE CLASES")
print("=" * 90)

print(f"\n{imbalance_metrics['severity_emoji']} CLASIFICACIÓN DE DESBALANCE: {imbalance_metrics['severity']}")
print("=" * 60)

print(f"\n📊 MÉTRICAS PRINCIPALES:")
print(f"   • Imbalance Ratio (IR):           {imbalance_metrics['imbalance_ratio']:.2f}")
print(f"   • Coefficient of Variation:       {imbalance_metrics['cv']:.3f}")
print(f"   • Gini Coefficient:               {imbalance_metrics['gini']:.3f}")
print(f"   • Shannon Entropy:                {imbalance_metrics['shannon_entropy']:.3f}")
print(f"   • Normalized Entropy:             {imbalance_metrics['normalized_entropy']:.3f}")

print(f"\n🧮 ESTADÍSTICAS DESCRIPTIVAS:")
print(f"   • Clase con más muestras:         {class_names[imbalance_metrics['total_counts'].index(imbalance_metrics['max_count'])]} ({imbalance_metrics['max_count']} muestras)")
print(f"   • Clase con menos muestras:       {class_names[imbalance_metrics['total_counts'].index(imbalance_metrics['min_count'])]} ({imbalance_metrics['min_count']} muestras)")
print(f"   • Promedio de muestras:           {imbalance_metrics['mean_count']:.1f}")
print(f"   • Desviación estándar:            {imbalance_metrics['std_count']:.1f}")
print(f"   • Diferencia max-min:             {imbalance_metrics['max_count'] - imbalance_metrics['min_count']} muestras")

print(f"\n📈 TEST ESTADÍSTICO (Chi-square):")
print(f"   • Chi-square statistic:           {imbalance_metrics['chi2_stat']:.2f}")
print(f"   • P-value:                        {imbalance_metrics['chi2_p_value']:.2e}")
if imbalance_metrics['chi2_p_value'] < 0.05:
    print("   • Resultado: ❌ RECHAZAMOS hipótesis de uniformidad (p < 0.05)")
else:
    print("   • Resultado: ✅ NO rechazamos hipótesis de uniformidad (p >= 0.05)")

print(f"\n📋 TABLA DETALLADA POR CLASE:")
print("=" * 90)
print(df_balance.to_string(index=False, float_format='%.2f'))

print(f"\n💡 INTERPRETACIÓN DE MÉTRICAS:")
print("=" * 60)
print(f"📌 Imbalance Ratio (IR = {imbalance_metrics['imbalance_ratio']:.2f}):")
if imbalance_metrics['imbalance_ratio'] <= 1.5:
    print("   ✅ Excelente: Dataset muy balanceado")
elif imbalance_metrics['imbalance_ratio'] <= 3:
    print("   🟡 Bueno: Ligero desbalance, manejable")
elif imbalance_metrics['imbalance_ratio'] <= 10:
    print("   🟠 Moderado: Considerar técnicas de balanceo")
else:
    print("   🔴 Severo: Requiere técnicas de balanceo obligatorias")

print(f"\n📌 Gini Coefficient ({imbalance_metrics['gini']:.3f}):")
if imbalance_metrics['gini'] < 0.1:
    print("   ✅ Distribución muy equitativa")
elif imbalance_metrics['gini'] < 0.2:
    print("   🟡 Distribución moderadamente equitativa")
else:
    print("   🔴 Distribución desigual")

print(f"\n📌 Normalized Entropy ({imbalance_metrics['normalized_entropy']:.3f}):")
if imbalance_metrics['normalized_entropy'] > 0.95:
    print("   ✅ Muy alta diversidad (cercano al balance perfecto)")
elif imbalance_metrics['normalized_entropy'] > 0.85:
    print("   🟡 Alta diversidad")
else:
    print("   🔴 Baja diversidad (concentrado en pocas clases)")

print(f"\n🎯 RECOMENDACIONES ESPECÍFICAS:")
print("=" * 60)

if imbalance_metrics['imbalance_ratio'] > 3:
    print("✅ TÉCNICAS DE BALANCEO RECOMENDADAS:")
    print("   • Class weights en la función de pérdida")
    print("   • Oversampling de clases minoritarias (SMOTE, ADASYN)")
    print("   • Undersampling de clases mayoritarias")
    print("   • Data augmentation focalizada en clases minoritarias")

if imbalance_metrics['cv'] > 0.3:
    print("✅ ESTRATEGIAS DE VALIDACIÓN:")
    print("   • Usar Stratified K-Fold cross-validation")
    print("   • Métricas balanceadas (F1-macro, AUC-macro)")
    print("   • Considerar Cohen's Kappa como métrica adicional")

if imbalance_metrics['chi2_p_value'] < 0.001:
    print("✅ MONITOREO ESPECIAL:")
    print("   • Revisar performance por clase individualmente")
    print("   • Matriz de confusión detallada")
    print("   • Métricas de precision/recall por clase")
```

```python
# Análisis de Calidad de Imágenes y Canales RGB

from collections import Counter
import seaborn as sns

def analyze_image_quality_and_colors(dataset, sample_size=1000):
    """
    Analiza la calidad de las imágenes y distribución de colores RGB
    """
    print("Analizando calidad de imágenes y colores RGB...")
    print("=" * 60)
    
    # Listas para almacenar estadísticas
    dimensions = []
    resolutions = []
    aspect_ratios = []
    rgb_stats = {'red': [], 'green': [], 'blue': []}
    unique_colors_per_image = []
    brightness_levels = []
    contrast_levels = []
    
    # Tomamos una muestra del dataset para análisis
    sample_indices = np.random.choice(len(dataset), min(sample_size, len(dataset)), replace=False)
    
    for i, idx in enumerate(sample_indices):
        if i % 200 == 0:
            print(f"Procesando imagen {i+1}/{len(sample_indices)}...")
        
        image, _ = dataset[idx]
        
        # Convertimos a numpy array si es tensor
        if hasattr(image, 'numpy'):
            img_array = image.numpy()
        else:
            img_array = np.array(image)
        
        # Aseguramos que esté en formato (H, W, C)
        if img_array.shape[0] == 3:  # Si está en formato (C, H, W)
            img_array = np.transpose(img_array, (1, 2, 0))
        
        # Información básica de la imagen
        height, width, channels = img_array.shape
        dimensions.append((height, width))
        resolutions.append(height * width)
        aspect_ratios.append(width / height)
        
        # Análisis de canales RGB (normalizamos a 0-255 si es necesario)
        if img_array.max() <= 1.0:
            img_array = (img_array * 255).astype(np.uint8)
        
        # Estadísticas por canal
        rgb_stats['red'].append({
            'mean': np.mean(img_array[:, :, 0]),
            'std': np.std(img_array[:, :, 0]),
            'min': np.min(img_array[:, :, 0]),
            'max': np.max(img_array[:, :, 0])
        })
        
        rgb_stats['green'].append({
            'mean': np.mean(img_array[:, :, 1]),
            'std': np.std(img_array[:, :, 1]),
            'min': np.min(img_array[:, :, 1]),
            'max': np.max(img_array[:, :, 1])
        })
        
        rgb_stats['blue'].append({
            'mean': np.mean(img_array[:, :, 2]),
            'std': np.std(img_array[:, :, 2]),
            'min': np.min(img_array[:, :, 2]),
            'max': np.max(img_array[:, :, 2])
        })
        
        # Cantidad de colores únicos
        img_flat = img_array.reshape(-1, 3)
        unique_colors = len(np.unique(img_flat.view(np.dtype((np.void, img_flat.dtype.itemsize * 3)))))
        unique_colors_per_image.append(unique_colors)
        
        # Brightness (promedio de todos los píxeles)
        brightness = np.mean(img_array)
        brightness_levels.append(brightness)
        
        # Contrast (desviación estándar)
        contrast = np.std(img_array)
        contrast_levels.append(contrast)
    
    return {
        'dimensions': dimensions,
        'resolutions': resolutions,
        'aspect_ratios': aspect_ratios,
        'rgb_stats': rgb_stats,
        'unique_colors': unique_colors_per_image,
        'brightness': brightness_levels,
        'contrast': contrast_levels
    }
```

Ratio de Desbalance (Imbalance Ratio): Se calcula dividiendo el número de instancias de la clase mayoritaria entre el número de instancias de la clase minoritaria. Un ratio alto indica un gran desbalance.

Ratio = Clase minoritaria / Clase mayoritaria

Por ejemplo, con 9500 y 500 instancias, el ratio sería 9500/500=19. Un valor de 1 indicaría un dataset perfectamente balanceado.
